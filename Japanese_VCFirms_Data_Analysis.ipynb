## DATA PREPROCESSING

### Null Value Heading

```python
# check the count of null records present in the data frame
df_japanese_vcfirms.isnull().sum()
```

**INTERPRETATION**
 - The categorical columns (startup_id, company_name, sector) have no null records.

### Drop columns with all unique values

```python
# Drop columns where all values are unique (like IDs)
unique_value_cols = [col for col in df_japanese_vcfirms.columns if df_japanese_vcfirms[col].nunique() == len(df_japanese_vcfirms)]
print("Columns with all unique values:", unique_value_cols)
df_japanese_vcfirms.drop(columns=unique_value_cols, inplace=True)
```

```python
df_japanese_vcfirms.columns
```

**INTERPRETATION**
 - The column startup_id had all unique values and is now dropped. This is because that column would not help the model as there is no variability.

### Drop columns with all same values

```python
# Drop columns with only one unique value
single_value_cols = [col for col in df_japanese_vcfirms.columns if df_japanese_vcfirms[col].nunique() == 1]
print("Columns with single repeated value:", single_value_cols)
df_japanese_vcfirms.drop(columns=single_value_cols, inplace=True)
```

```python
df_japanese_vcfirms.columns
```

**INTERPRETATION**
 - There is no column with all same values in the dataset.

### Handling Null Values

```python
# Drop columns with too many nulls (e.g., >40% of dataset size)
threshold = 0.4 * len(df_japanese_vcfirms)
df_japanese_vcfirms = df_japanese_vcfirms.dropna(thresh=threshold, axis=1)

# For remaining nulls, impute
for col in df_japanese_vcfirms.columns:
    if df_japanese_vcfirms[col].isnull().sum() > 0:
        if df_japanese_vcfirms[col].dtype == 'object':
            df_japanese_vcfirms[col].fillna(df_japanese_vcfirms[col].mode()[0], inplace=True)
        else:
            df_japanese_vcfirms[col].fillna(df_japanese_vcfirms[col].median(), inplace=True)
```

```python
# check all the columns
df_japanese_vcfirms.columns
```

```python
# check the count of null values
df_japanese_vcfirms.isnull().sum()
```

**INTERPRETATION**
 - There are no null records in the dataset.

```python
# Decided to take Green_Cover_Percentage as the target variable(dependent variable)
target = df_japanese_vcfirms.success_score

# After storing this variable into target, we have to remove it from the main data frame(independent variables)
df_japanese_vcfirms = df_japanese_vcfirms.drop('success_score', axis = 1)
```

## Prepare the data for Machine Learning

```python
# Seggregate the data based on the data types
df_num = df_japanese_vcfirms.select_dtypes(include = 'number')
```

```python
# Make separate data frame for categorical variables
df_cat = df_japanese_vcfirms.select_dtypes(include = 'object')
```

## Perform encoding on categorical variables

```python
# Perform encoding(one hot encoding)
df_encoded = pd.get_dummies(df_cat)
```

```python
# Check the encoded data
df_encoded.head()
```

### Scale the numerical variables

```python
# import the standard scalar subpackage for performing the scaling
from sklearn.preprocessing import StandardScaler
```

```python
# Initiate the standard scaler
ss = StandardScaler()
```

```python
# Transform the numerical variables
scale = ss.fit_transform(df_num)

# Check the transformed data
scale
```

```python
# Create a new data frame for scaled data
df_scaled = pd.DataFrame(scale, columns = df_num.columns)
```

```python
# Check the scaled data frame
df_scaled.head()
```

### Concatenate both the encoded and scaled data frame

```python
# Concatenation
X = pd.concat([df_encoded, df_scaled], axis = 1)

# Check the concatenated data
X.head()
```

## Split the data in Training Set and Testing Set 

```python
# import the train test split
from sklearn.model_selection import train_test_split
```

```python
# Create xtrain, xtest, ytrain, ytest
X_train, X_test, y_train, y_test = train_test_split(X, target, test_size = 0.20, random_state = 42)
```

```python
# Authetic check training set
print(X_train.shape)

print(y_train.shape)
```

```python
# Authentic check for testing set
print(X_test.shape)

print(y_test.shape)
```

## Linear Regression

```python
# import Linear Regression Algorithm and evaluation matrix 
from sklearn.linear_model import LinearRegression

# Here we are using MSE as evaluation matrix
from sklearn.metrics import mean_squared_error
```

```python
# Create a Linear Regression Model
model = LinearRegression()
```

```python
# Fit the model to the training data
model.fit(X_train, y_train)
```

```python
# Predicting on the testing data
predictions = model.predict(X_test)
```

```python
# Calculate mean squared error
mse = mean_squared_error(y_test, predictions)
print("Mean Squared Error:", mse)
```

## Ridge Model

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

# Assuming X_train, X_test, y_train, y_test are already defined from train_test_split

# Create a Ridge regression model
ridge_model = Ridge(alpha=1.0)  # alpha is the regularization strength

# Fit the model to the training data
ridge_model.fit(X_train, y_train)

# Predict on the testing data
predictions = ridge_model.predict(X_test)

# Calculate mean squared error
mse = mean_squared_error(y_test, predictions)
print("Mean Squared Error:", mse)
```

## Lasso Model

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error

# Assuming X_train, X_test, y_train, y_test are already defined from train_test_split

# Create a Lasso regression model
lasso_model = Lasso(alpha=1.0)  # alpha is the regularization strength

# Fit the model to the training data
lasso_model.fit(X_train, y_train)

# Predict on the testing data
predictions = lasso_model.predict(X_test)

# Calculate mean squared error
mse = mean_squared_error(y_test, predictions)
print("Mean Squared Error:", mse)
```

## Decision Tree Regressor
